```python
#Find the most frequent value and their indice(s)
x = np.array([1,2,3,4,5,1,2,1,1,1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

def most_frequent_value_and_indices(array):
    values, counts = np.unique(array, return_counts=True)
    maximum_count = np.max(counts)
    most_frequent_values = values[counts == maximum_count]
    result = []
    for value in most_frequent_values:
        indices = np.where(array == value)[0]
        result.append((value, indices))
    return maximum_count, result
x_count, x_result = most_frequent_value_and_indices(x)

print("Array x:", x)
print("Maximum frequency in x:", x_count)

for value, indices in x_result:
    print("Most frequent value:", value)
    print("Indices:", indices)


y_count, y_result = most_frequent_value_and_indices(y)

print("\nArray y:", y)
print("Maximum frequency in y:", y_count)

for value, indices in y_result:
    print("Most frequent value:", value)
    print("Indices:", indices)
```

Q.4 Write program to create an 1-D NumPy array named <<Your Name>> with evenly
spaced 25 numbers from 10 to 100 using linspace(). Print the dimensions of the array,
shape, total elements, the data type of each element and total number of bytes consumed
by the array. Find the transpose of this array using reshape() attribute. Can we do the same
with T attribute?

```python
print("\nQ4. Array information using linspace()")

tanay_bewtra = np.linspace(10, 100, 25)
print("Array tanay_bewtra:")
print(tanay_bewtra)
```

```python
print("dimensions:", tanay_bewtra.ndim)
print("Shape:", tanay_bewtra.shape)
print("Total number:", tanay_bewtra.size)
print("Data type:", tanay_bewtra.dtype)
print("Total bytes consumed:", tanay_bewtra.nbytes)
```

```python
transpose_using_reshape = tanay_bewtra.reshape(1, 25).reshape(25, 1)
print("\nTranspose using reshape():")
print(transpose_using_reshape)
```

```python
transpose_using_T_1d = tanay_bewtra.T
print("\nUsing .T directly on the 1-D array:")
print(transpose_using_T_1d)
```

Q5. Create a 2-D Array of three rows and four columns, named ucs420_<your_name>>
with following values – 10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35. Compute the mean,
median, max, min, unique elements. Reshape the array to four rows and three columns and
name it as reshaped_ ucs420_<your_name>>. Resize the array to two rows and three
columns and name it as resized_ ucs420_<your_name>>.

```python
ucs420_tanay_bewtra = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])
```

```python
print("Mean: ",np.mean(ucs420_tanay_bewtra))
```

```python
print("Median:", np.median(ucs420_tanay_bewtra))
print("Maximum value:", np.max(ucs420_tanay_bewtra))
print("Minimum value:", np.min(ucs420_tanay_bewtra))
print("Unique elements:", np.unique(ucs420_tanay_bewtra))
```

```python
reshaped_ucs420_tanay_bewtra = ucs420_tanay_bewtra.reshape(4, 3)
print("\nReshaped array:")
print(reshaped_ucs420_tanay_bewtra)
```

```python
resized_ucs420_tanay_bewtra = np.resize(ucs420_tanay_bewtra, (2, 3))
print("\nResized array:")
print(resized_ucs420_tanay_bewtra)
```
