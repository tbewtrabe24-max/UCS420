Q1. Consider the following sensor readings:
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

```python
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
print(temperature)
```

```python
#a. Add 2°C to every reading using vectorization.
corrected_temperature = temperature + 2

print("\nCorrected readings after adding +2°C:")
print(corrected_temperature)
```

```python
#b. Convert all temperatures from Celsius to Fahrenheit
fahrenheit_temperature = (temperature * 9/5) + 32

print("\nTemperatures in Fahrenheit:")
print(fahrenheit_temperature)
```

```python
#c. Identify all readings greater than 32°C using Boolean indexing.
high_temperature = temperature[temperature > 32]

print("\nTemperatures greater than 32°C:")
print(high_temperature)
```

```python
#d. Count how many readings exceed 32°C.
count_high_temperature = np.sum(temperature > 32)

print("\nNumber of temperatures greater than 32°C:", count_high_temperature)
```

```python
#e. Explain how vectorization and Boolean indexing can make data processing more efficient
#   than explicitly iterating through every value.
print("""
Vectorization applies an operation to an entire NumPy array at once.
For example, temperature + 2 adds 2 to every sensor value without a
Python for loop. NumPy performs these array operations efficiently in
optimized low-level code.

Boolean indexing first creates a True/False condition, such as
temperature > 32, and then directly selects only the values for which
the condition is True. This is concise, readable, and generally faster
than checking every value manually using a loop.
""")
```

Q2. The following matrix represents the number of daily steps recorded by four users
over three days:

```python
steps = np.array([
[5000, 6200, 7100],
[8000, 7500, 9000],
[4500, 5100, 4800],
[9000, 8500, 9500]])
```

a. Calculate the total steps recorded.

b. Calculate the mean number of steps.

c. Find the maximum and minimum values.

d. Calculate the total steps for each day using axis=0.

e. Calculate the total steps for each user using axis=1.

f. Find the user/day position containing the maximum number of steps using argmax().

```python
#a
print(steps.sum())
```

```python
#b
print(np.mean(steps))
```

```python
print(np.max(steps))
print(np.min(steps))
```

```python
#d
daily_total_steps = np.sum(steps, axis=0)
print("Total steps for each day:", daily_total_steps)
```

```python
user_total_steps = np.sum(steps, axis=1)
print("Total steps for each user:", user_total_steps)
```

```python
flat_max_index = np.argmax(steps)
max_position = np.unravel_index(flat_max_index, steps.shape)
print("User/day position with the maximum number of steps:", max_position)
```

```python
matrix_3x4 = np.arange(1, 13).reshape(3, 4)

print("3 x 4 Matrix:")
print(matrix_3x4)
```

```python
first_row = matrix_3x4[0]
last_row = matrix_3x4[-1]
second_column = matrix_3x4[:, 1]

# Rows 1–2 and columns 2–3 in normal human counting:
# Row indices 0:2 and column indices 1:3
rows_1_to_2_columns_2_to_3 = matrix_3x4[1:3, 2:4]


print("First row:", first_row)
print("Last row:", last_row)
print("Second column:", second_column)

print("\nElements from rows 1-2 and columns 2-3:")
print(rows_1_to_2_columns_2_to_3)
```

```python
flattened_array = matrix_3x4.flatten()
ravelled_array = matrix_3x4.ravel()

print("Using flatten():", flattened_array)
print("Using ravel():", ravelled_array)
```

```python
print("Matrix before modifying ravelled array:")
print(matrix_3x4)

ravelled_array[0] = 999

print("\nRavelled array after modification:")
print(ravelled_array)

print("\nOriginal matrix after modifying ravelled array:")
print(matrix_3x4)
```

```python
# Changing ravelled_array[0] also changes matrix_3x4[0, 0].
# This happens because ravel() generally returns a view of the original
# array when NumPy can do so without copying data.
```

```python
matrix_3x4[0, 0] = 1
flattened_array = matrix_3x4.flatten()
print("Matrix before modifying flattened array:")
print(matrix_3x4)
flattened_array[0] = 500
print("\nFlattened array after modification:")
print(flattened_array)
print("\nOriginal matrix after modifying flattened array:")
print(matrix_3x4)
```

```python
# Changing flattened_array[0] does not change matrix_3x4[0, 0].
# This is because flatten() creates and returns a copy of the original
# array rather than a view.
```

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
], dtype=float)

# Assistance scores
y = np.array([40, 65, 30, 85], dtype=float)

print("Feature Matrix X:")
print(X)

print("\nTarget vector y:")
print(y)
```

```python
print("Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)
```

```python
X_transpose = X.T

print("Transpose of X (X.T):")
print(X_transpose)
```

```python
XTX = X.T @ X

print("X.T @ X:")
print(XTX)
```

```python
inverse_XTX = np.linalg.inv(XTX)

print("Inverse of (X.T @ X):")
print(inverse_XTX)
```

```python
beta = np.linalg.inv(X.T @ X) @ X.T @ y

print("OLS coefficients (beta):")
print(beta)

print("\nSleep-hours coefficient:", beta[0])
print("Activity-level coefficient:", beta[1])
print("Stress-level coefficient:", beta[2])
```

```python
print(f"""
Interpretation of coefficients:

1. Sleep-hours coefficient = {beta[0]:.2f}
   Holding other features constant, an increase of one sleep hour is
   associated with approximately {beta[0]:.2f} change in assistance score.

2. Activity-level coefficient = {beta[1]:.2f}
   Holding other features constant, a one-unit increase in activity level
   is associated with approximately {beta[1]:.2f} change in assistance score.

3. Stress-level coefficient = {beta[2]:.2f}
   Holding other features constant, a one-unit increase in stress level is
   associated with approximately {beta[2]:.2f} change in assistance score.

Note: This model does not contain an intercept/bias column of ones because
the assignment provides X with only three feature columns.
""")
```

```python
new_user = np.array([5, 40, 7], dtype=float)

predicted_assistance_score = new_user @ beta

print("New user [Sleep Hours, Activity Level, Stress Level]:")
print(new_user)

print("\nPredicted assistance score:")
print(predicted_assistance_score)
```
