# Assignment 5.3 ADD ALL PRIME NUMBERS FROM 1 TO N AND N IS GIVEN BY THE USER

```python
n = int(input("Enter a number: "))
prime_sum = 0

for num in range(2, n + 1):
    is_prime = True
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        prime_sum += num

print(prime_sum)
```

# Assignment 6.1 ADD ALL THE ODD NUMBERS FROM 1 TO N AND N IS GIVEN BY THE USER

```python
def add_odd(n):
  sum = 0
  for i in range(1,n+1):
    if i%2!=0:
      sum += i
  return sum
add_odd(10)
```

# Assignment 6.2 ADD ALL PRIME NUMBER USING FUNCTION AND N IS GIVEN BY USER

```python
def add_prime(n):
  sum_primes = 0
  for num in range(2, n + 1):
      is_prime = True
      for j in range(2, int(num**0.5) + 1):
          if num % j == 0:
              is_prime = False
              break
      if is_prime:
          sum_primes += num
  return sum_primes
print(f"Sum of primes up to 10: {add_prime(10)}")
```
